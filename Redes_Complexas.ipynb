```python
import networkx as nx
import matplotlib.pyplot as plt

# Número de nós
num_nodes = 20

# Criação do grafo completo
G = nx.complete_graph(num_nodes)

# Definir layout para uma visualização organizada
pos = nx.spring_layout(G)

# Plotar o grafo
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_size=700, node_color="red", font_size=15, font_weight="bold", edge_color="gray")

plt.title("Rede Completa com 10 Nós")
plt.show()
```

```python
# Plotar histograma do grau da rede
graus = [degree for node, degree in G.degree()]
plt.figure(figsize=(8, 6))
plt.hist(graus, bins=range(min(graus), max(graus) + 2), edgecolor='black', color='lightgreen')
plt.title("Histograma do Grau da Rede")
plt.xlabel("Grau")
plt.ylabel("Frequência")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import random

# Número de nós
num_nodes = 50
probabilidade_conexao = 0.2

# Criação do grafo com probabilidade fixa de conexão
G = nx.Graph()
G.add_nodes_from(range(num_nodes))

# Adicionar arestas com probabilidade Pe
for i in range(num_nodes):
    for j in range(i + 1, num_nodes):
        if random.random() < probabilidade_conexao:
            G.add_edge(i, j)

# Definir layout para uma visualização organizada
pos = nx.spring_layout(G)

# Plotar o grafo
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_size=700, node_color="skyblue", font_size=15, font_weight="bold", edge_color="gray")
plt.title("Rede com Probabilidade de Conexão Pe=0.5")
plt.show()

# Plotar histograma do grau da rede
graus = [degree for node, degree in G.degree()]
plt.figure(figsize=(8, 6))
plt.hist(graus, bins=range(min(graus), max(graus) + 2), edgecolor='black', color='lightgreen')
plt.title("Histograma do Grau da Rede")
plt.xlabel("Grau")
plt.ylabel("Frequência")
plt.show()
```
